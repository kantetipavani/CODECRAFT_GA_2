```python
import torch
from diffusers import StableDiffusionPipeline
import gradio as gr

device = "cuda" if torch.cuda.is_available() else "cpu"

model_id = "prompthero/openjourney-v4"

pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to(device)

def generate_image(prompt):
    try:
        image = pipe(prompt, num_inference_steps=25, guidance_scale=7.5).images[0]
        image.save("/content/generated.png")
        return image, "✅ Image generated successfully!"
    except Exception as e:
        return None, f"⚠️ Error: {e}"

interface = gr.Interface(
    fn=generate_image,
    inputs=gr.Textbox(label="Enter your prompt", placeholder="e.g. A futuristic city with flying cars"),
    outputs=[gr.Image(label="Generated Image"), gr.Textbox(label="Status", interactive=False)],
    title="🖼️ Text-to-Image Generator (OpenJourney v4)",
    description="Generate AI art from text using Stable Diffusion. No token required.",
    theme="soft"
)

interface.launch(debug=True)
```
